# wrangle / household / income

- Outputs `household_out`, a scaffold name until the signature is defined.
- Gives the income that does not come from drawing down the portfolio, by source and age to 2070.
- Covers salary, dividends, interest, Age Pension and superannuation.
- Lets a withdrawal fund only the rest, for `portfolio/withdrawal` and `portfolio/profile`.

**Data**
- Personal data is git-ignored in `data/` and `project/`, with synthetic placeholders committed.

- `validation` and `backtest` are later years, held back for acceptance.
- Salary, dividends and interest already land in the `Bank` workbook categories.
- Super and pension statements are on Drive under `Wrangle/Super` and `Wrangle/Pensions`.

**Status**
