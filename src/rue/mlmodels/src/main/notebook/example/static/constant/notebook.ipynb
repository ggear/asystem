# example / static / constant

**Intent**
- Gives `static_out` as the constant 7, whatever `static_in` is, and must:
  - Be the smallest complete project, written, packaged and replayed with no fitting.
  - Stay a fixture, tested by `fab unittest` and `fab systest` and never released.

**Data**
- `golden` is the one input and expected output, then a request that must give a signature error.
- Every other split is unused.

## Setup ([6.7.1](https://github.com/ggear/asystem/blob/master/src/rue/mlmodels/src/build/resources/doc/model.md#671-setup))

```python
from mlmodels import setup_workspace

ws = setup_workspace("example/static/constant")
```

## Define ([6.7.6](https://github.com/ggear/asystem/blob/master/src/rue/mlmodels/src/build/resources/doc/model.md#676-define))

```python
from mlmodels import define_model

CONSTANTS = {"static_out": 7}
define_model(ws, model=1000, flavour="static", source=CONSTANTS, label="The constant 7")
```

## Package ([6.7.8](https://github.com/ggear/asystem/blob/master/src/rue/mlmodels/src/build/resources/doc/model.md#678-package))

```python
from mlmodels import package_model

package_model(ws, 1000, {"static_in": [1.0, None], "expect": ["ok", "error"]}, target={"kind": "embedded", "language": "rust"})
```
