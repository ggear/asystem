# finance / economy / inflation

Inflation to 2070 as three constants: low, base and high.

| Term | In This Notebook |
|---|---|
| **Intent** | Give one annual inflation rate per scenario to 2070 for the economy, separate from `household/costs`. Placeholder: the base is the mean annual inflation and the low and high are one standard deviation either side |
| **Project Flavour** | `static`, served by `mlserver` (`model.md` 3.2) |
| **Data Splits** | `training` ten years of annual inflation, synthetic around the 3.24 percent 40 year mean until the `interest` table is read; `golden` the exact request that becomes `golden.json` |

## Setup

- **Stage:** none, preparation for the stages below.
- **Intent:** put `workbench` on the path, load `env`, and declare the model name, target, dataset names and signature names as constants (`model.md` 6.6).

```python
import json

from workbench.data import ensure_data, golden_data
from workbench.env import init_env
from workbench.project import provenance, record_project, scratch_project_dir

env = init_env()

WORKSPACE = "finance/economy/inflation"
MODEL_NAME = "finance_economy_inflation"
TARGETS = [{"kind": "served"}]
TRAINING = "economy_inflation_training_2026-11-05_2026-11-05"
GOLDEN = "economy_inflation_golden"
INPUT = "finance_economy_inflation_in"
OUTPUTS = ["finance_economy_inflation_annual_low_pct", "finance_economy_inflation_annual_base_pct", "finance_economy_inflation_annual_high_pct"]
```

## Capture Data

- **Stage:** 1, capture data (`model.md` 7.1).
- **Intent:** write the `training` file only when it is absent (`model.md` 6.5). The source is synthetic until the real series is loaded, so no query runs.

```python
DEVIATIONS = [-1.2, 0.8, -0.4, 1.0, -0.6, 0.2, 0.6, -1.0, 0.4, 0.2]

training_pl, training_path = ensure_data(WORKSPACE, "training", TRAINING, lambda: {"year": list(range(2016, 2026)), "inflation_pct": [round(3.24 + deviation, 4) for deviation in DEVIATIONS]})
print("training", training_pl.height)
```

## Fit

- **Stage:** 3, fit.
- **Intent:** take the mean and the population standard deviation of annual inflation in `training` only (R6). The scenarios are one standard deviation either side until the mean reverting regime model replaces them (see the plan).

```python
BASE = round(float(training_pl["inflation_pct"].mean()), 2)
SPREAD = round(float(training_pl["inflation_pct"].std(ddof=0)), 2)
CONSTANTS = dict(zip(OUTPUTS, [round(BASE - SPREAD, 2), BASE, round(BASE + SPREAD, 2)], strict=True))
print("constants", CONSTANTS)
assert list(CONSTANTS.values()) == [2.52, 3.24, 3.96], "the fit did not recover the mean and spread"
```

## Package

- **Stage:** 5, package.
- **Intent:** write the golden rows, build the signature, write a `static` project whose constants are the fitted values, verify it and replay the golden through the library loader (R1, R3).

```python
from workbench.model import build_signature, package_project, register_preview, signature_input, signature_output
from workbench.replay import require_replay

golden_pl, golden_path = golden_data(WORKSPACE, GOLDEN, lambda: {INPUT: [1.0, None], "expect": ["ok", "error"]})

signature = build_signature(
    MODEL_NAME,
    [signature_input(INPUT, "none", "Unused input.")],
    [
        signature_output("finance_economy_inflation_annual_low_pct", "percent", "Annual inflation rate, percent per year, low scenario.", datatype="FP64"),
        signature_output("finance_economy_inflation_annual_base_pct", "percent", "Annual inflation rate, percent per year, base scenario.", datatype="FP64"),
        signature_output("finance_economy_inflation_annual_high_pct", "percent", "Annual inflation rate, percent per year, high scenario.", datatype="FP64"),
    ],
)
project_dir = scratch_project_dir(WORKSPACE)
manifest = package_project(
    project_dir,
    MODEL_NAME,
    "static",
    signature,
    golden_pl.iter_rows(named=True),
    provenance(
        "manual",
        "synthetic annual inflation, training split",
        [training_path, golden_path],
        "mean and standard deviation of annual inflation",
    ),
    constants=CONSTANTS,
)
print(json.dumps(manifest, indent=2))

require_replay(project_dir)
metadata = record_project(WORKSPACE, project_dir, TARGETS)
print("replayed the golden sequences exactly through the shipped loader")
```

## Register

- **Stage:** 6, register.
- **Intent:** register the verified project with the alias `preview`. A person sets `release` after review and the parity gate (R5).

```python
version = register_preview(project_dir, MODEL_NAME, env["MLFLOW_TRACKING_URI"], TARGETS, metadata["snapshot_version"])
print("registered version", version)
```
